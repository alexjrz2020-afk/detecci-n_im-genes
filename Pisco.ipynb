```python
from ultralytics import YOLO
from PIL import Image
import matplotlib.pyplot as plt

# 1️⃣ Cargar modelo entrenado
model = YOLO("/content/runs/detect/pisco_v2_finetune/weights/best.pt")

# 2️⃣ Cargar imagen de prueba
img_path = "/content/Pisco_peru.jpg"  # Cambia a cualquier imagen
img = Image.open(img_path)

# 3️⃣ Hacer predicción
results = model(img)

# 4️⃣ Filtrar solo pisco (clase 7 según tu dataset)
pisco_results = []
for i, cls in enumerate(results[0].boxes.cls):  # <-- cambio aquí
    if int(cls) == 7:  # Clase "pisco"
        pisco_results.append(results[0].boxes[i])

# 5️⃣ Mostrar imagen con detecciones solo de pisco
img_out = results[0].plot(boxes=pisco_results)
plt.figure(figsize=(8,8))
plt.imshow(img_out)
plt.axis("off")
plt.show()

# 6️⃣ (Opcional) imprimir las coordenadas y la confianza
for box in pisco_results:
    print(f"Coordenadas: {box.xyxy}, Confianza: {box.conf}")
```

```python
from ultralytics import YOLO

# Cargar modelo entrenado
model = YOLO("/content/runs/detect/pisco_v2_finetune/weights/best.pt")

# Evaluación en validación
results = model.val()  # usa dataset valid del data.yaml

# Métricas globales
metrics = results.results_dict
print("Métricas globales:")
print(metrics)

# mAP por clase
print("\nmAP por clase:")
for i, cls_name in results.names.items():
    print(f"{cls_name}: {results.maps[i]:.3f}")

# Resumen bonito
print("\nResumen por clase:")
results.summary()
```
